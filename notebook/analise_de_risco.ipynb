```python
from prophet import Prophet
import pandas as pd
```

<p style="font-size: 20px; color:#66ffff; text-align: center">
     Padronização de coluna - 
     Column Normalization
</p>

```python
def padrao_colunas(df,colunas):
    for coluna in colunas:
        df[coluna] = (df[coluna].astype(str).str.replace(".","",regex=False).str.replace(",", ".",regex=False)) 
        df[coluna] = pd.to_numeric(df[coluna],errors="coerce")
    return df
```

<p style="font-size: 20px; color:#66ffff; text-align: center">
     Importação dos dados Consolidadado para Predição - 
     Importing Consolidated Data for Prediction
</p>

```python
df = pd.read_excel(r"C:\Users\admin\Documents\Project_Capstone\source\data\dados_consolidados.xlsx")
```

```python
colunas_convertidas = ["receita_do_mes","despesa_total","lucro_bruto","y","despesas_acumulada","lucro_acumulado"]
df = padrao_colunas(df,colunas_convertidas)
```

<p style="font-size: 20px; color:#66ffff; text-align: center">
     Aplicação do modelo Prophet para predição - 
     Applying Prophet model for forecasting
</p>

```python
def predicoes(df,limite_risco,meses=6):
    df_predicao = df.copy()
    df_predicao.columns = ["ds", "y"]

    
    modelo = Prophet()
    modelo.fit(df_predicao)


    futuro = modelo.make_future_dataframe(periods=meses)
    previsao = modelo.predict(futuro)

    risco = limite_risco
    previsao["alerta_risco"] = previsao["yhat"].apply(
    lambda valor:"risco" if valor < risco else "ok"
    ) 
    return previsao
```

```python
predicao_lucro = predicoes(df[["ds","y"]],limite_risco=42000,meses=12)
predicao_despesas_acumulada = predicoes(df[["ds","despesas_acumulada"]],50000)
```
